```python
def sigmavrelic(xf, mchi):
    return 0.1*xf*1e-8/(np.sqrt(gstarfun(mchi))*OmegaDM) 

def sigmavrelicconst(xf, mchi):
    return 0.1*xf*1e-8/(np.sqrt(gstarfunconst(mchi))*OmegaDM) 

gchi = 2.0      # internal dof of chi

def xfreeze(mchi, x0=20.0, tol=1e-10, itmax=100):
    """Freeze-out xf = mchi/T_f, solved for rather than guessed.

    The freeze-out condition Gamma = H reads

        xf = ln[ 0.038 g Mpl mchi <sigma v>(xf) / sqrt(g_*(T_f) xf) ]

    which looks unusable on its own, since <sigma v> depends on the epsilon
    we are trying to find. But at the solution <sigma v> is by definition the
    value that reproduces the observed relic density, sigmavrelic(xf, mchi),
    so ma and epsilon drop out and this closes on xf alone.

    Iterated as a fixed point: the right-hand side grows like ln(xf), whose
    slope ~ 1/(2 xf) makes the map a strong contraction, so it converges in
    about ten passes and does not care about x0. Iterating also keeps the
    whole mchi array vectorised, which a root finder would not.
    """
    mchi = np.asarray(mchi, dtype=float)
    x = x0*np.ones_like(mchi)
    for _ in range(itmax):
        xnew = np.log(0.038*gchi*mpl*mchi*sigmavrelic(x, mchi)
                      / np.sqrt(gstarfun(mchi/x)*x))
        step = np.max(np.abs(xnew - x))
        x = xnew
        if step < tol:
            break
    else:
        print(f"xfreeze: no convergence to {tol:g} in {itmax} iterations")
    return x

def xfreezeconst(mchi, x0=20.0, tol=1e-10, itmax=100):
    """Freeze-out xf = mchi/T_f, solved for rather than guessed.

    The freeze-out condition Gamma = H reads

        xf = ln[ 0.038 g Mpl mchi <sigma v>(xf) / sqrt(g_*(T_f) xf) ]

    which looks unusable on its own, since <sigma v> depends on the epsilon
    we are trying to find. But at the solution <sigma v> is by definition the
    value that reproduces the observed relic density, sigmavrelic(xf, mchi),
    so ma and epsilon drop out and this closes on xf alone.

    Iterated as a fixed point: the right-hand side grows like ln(xf), whose
    slope ~ 1/(2 xf) makes the map a strong contraction, so it converges in
    about ten passes and does not care about x0. Iterating also keeps the
    whole mchi array vectorised, which a root finder would not.
    """
    mchi = np.asarray(mchi, dtype=float)
    x = x0*np.ones_like(mchi)
    for _ in range(itmax):
        xnew = np.log(0.038*gchi*mpl*mchi*sigmavrelicconst(x, mchi)
                      / np.sqrt(gstarfunconst(mchi/x)*x))
        step = np.max(np.abs(xnew - x))
        x = xnew
        if step < tol:
            break
    else:
        print(f"xfreeze: no convergence to {tol:g} in {itmax} iterations")
    return x


def vare(ma, alpha, alphaD, mchi):
    xf = xfreeze(mchi)
    #alphaarr = alpha*np.ones_like(ma)
    #alphaDarr = alphaD*np.ones_like(ma)
    return np.sqrt(sigmavrelic(xf, mchi)/(sigmavrel(1, ma, alpha, alphaD, mchi, xf)))

def vareconst(ma, alpha, alphaD, mchi):
    xf = xfreezeconst(mchi)
    #alphaarr = alpha*np.ones_like(ma)
    #alphaDarr = alphaD*np.ones_like(ma)
    return np.sqrt(sigmavrelicconst(xf, mchi)/(sigmavrel(1, ma, alpha, alphaD, mchi, xf)))

def varexfixed(ma, alpha, alphaD, mchi):
    xf = 22
    #alphaarr = alpha*np.ones_like(ma)
    #alphaDarr = alphaD*np.ones_like(ma)
    return np.sqrt(sigmavrelicconst(xf, mchi)/(sigmavrel(1, ma, alpha, alphaD, mchi, xf)))
```
